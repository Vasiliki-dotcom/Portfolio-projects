```python
url = 'https://raw.githubusercontent.com/fivethirtyeight/data/master/college-majors/recent-grads.csv'
df = pd.read_csv(url)
print(df.columns)
df.head()
#top_10 = df[['Major', 'Median']].sort_values(by='Median', ascending=False).head(10)
```

```python
import matplotlib.pyplot as plt
top_10 = df[['Major', 'Median']].sort_values(by='Median', ascending=False).head(15)
plt.figure(figsize=(10,6))
plt.barh(top_10['Major'], top_10['Median'], color='orange')
plt.xlabel('Salary in $')
plt.ylabel('Profession')
plt.title('Top 10 Highest Paying Degrees')
plt.gca().invert_yaxis()
plt.show()
```
